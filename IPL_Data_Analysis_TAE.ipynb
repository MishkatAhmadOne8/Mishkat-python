```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/ipl.csv')
```

```python
df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
df['winner'].value_counts().head(10)

```

```python
top_teams = df['winner'].value_counts().head(5)

plt.figure()
top_teams.plot(kind='bar')
plt.title('Top 5 Most Successful IPL Teams')
plt.xlabel('Team')
plt.ylabel('Number of Wins')
plt.show()
```
